## Oferecendo desconto para cliente com base na última compra

Suponha que você está trabalhando para uma empresa que deseja rastrear a atividade do cliente. Uma métrica que eles estão interessados é o tempo que passou desde a última transação do cliente. Se for muito tempo, eles podem oferecer um desconto para o cliente. Crie um script Python que mostra quanto tempo se passou desde a última compra do cliente. Se faz mais de 30 dias, mostre uma mensagem oferecendo um desconto para o cliente.

```python
from datetime import datetime

agora = datetime.now()
print(agora)

ultima_compra = datetime(2025, 5, 10)
print(ultima_compra)

diferenca_datas = agora - ultima_compra
print(diferenca_datas)
print(type(diferenca_datas))

dias = diferenca_datas.days
print(dias)

if dias > 30:
    print("Voce tem direito a uma promoçáo!!")
else:
    print("Voce comprou recente, obrigado!")
```

## Data e hora em diferentes fusos horários

Uma empresa tem escritórios em São Paulo, Nova York e Tóquio. Crie um script Python que mostra a data e hora atuais nesses três fusos horários. Exiba, também, se estes escritórios estão abertos ou fechados (9h às 17h).


```python
from datetime import datetime
from zoneinfo import ZoneInfo

data_atual = datetime.now()
print(data_atual)
```

```python
hora_sao_paulo = ZoneInfo("America/Sao_Paulo")
hora_nova_york = ZoneInfo("America/New_York")
hora_tokyo = ZoneInfo("Asia/Tokyo")

data_hora_sao_paulo = data_atual.astimezone(hora_sao_paulo)
data_hora_nova_york = data_atual.astimezone(hora_nova_york)
data_hora_tokyo = data_atual.astimezone(hora_tokyo)

print(f"A data e hora em São Paulo é {data_hora_sao_paulo}")
print(f"A data e hora em Nova York é {data_hora_nova_york}")
print(f"A data e hora em Tokyo é {data_hora_tokyo}")
```

```python
def verifica_horario(data_hora):
    if 9 <= data_hora.hour < 17:
        return "Aberto !"
    else:
        return "Fechado!"
    
print(f"O escritório de Sáo Paulo está {verifica_horario(data_hora_sao_paulo)}")
print(f"O escritório de Nova York está {verifica_horario(data_hora_nova_york)}")
print(f"O escritório de Tokyo está {verifica_horario(data_hora_tokyo)}")
```
