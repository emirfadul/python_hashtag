## Calculando a idade

Um usuário fornece sua data de nascimento no formato "dd/mm/aaaa". Crie um script Python que calcula a idade do usuário.


```python
from datetime import datetime

data_nasc = input("Informe sua data de nascimento (dd/mm/aaaa): ")
print(data_nasc)
```

```python
data_nasc = datetime.strptime(data_nasc, "%d/%m/%Y")
print(data_nasc)
```

```python
data_atual = datetime.now()
print(f"Hoje é {data_atual}")
```

```python
idade = data_atual.year - data_nasc.year
print(idade)
```
