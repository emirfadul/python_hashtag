```python
ingresso = float(input("qual valor ingresso: "))
publico = float(input("qual o publico: "))
montante = ingresso * publico

tx_cartao = 0.035
tx_gtickets = 0.015
total_tx = tx_cartao + tx_gtickets

tx_ticketeira = 0.15

print(f"*** OPERAÇÃO ***")
print(f"O Valor do ingresso R$ {ingresso:.2f}")
print(f"Estimativa de publico: {publico}")
print(f"Montante esperado R$ {montante:.2f}\n")

print(f"*** TAXAS ***")
print(f"Taxa de cartao R$ {(montante * tx_cartao):.2f}")
print(f"Taxa da G_Ticket R$ {montante * tx_gtickets}")
print(f"Total de taxas R$ {montante * total_tx}\n")

print(f"*** RESUMO ***")
print(f"Vendas esperadas R$ {montante}")
print(f"Valor ticketeira (15%) R$ {montante * 0.15}")
print(f"Valor liquido de repasse cliente R$ {montante - (montante * 0.15)}\n")

print(f"*** TICKETEIRA ***")
print(f"Valor liquido TICKETEIRA R$ {(montante * 0.15) - (montante * total_tx)}")



```
