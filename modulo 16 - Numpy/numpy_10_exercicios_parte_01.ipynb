## Exercício

Você é um gerente de vendas e tem os dados de vendas de três produtos diferentes (Produto A, Produto B, Produto C) para os últimos 5 dias em um array 2D NumPy. Cada linha do array representa um produto e cada coluna representa um dia. Seu trabalho é calcular as vendas totais para cada produto e para cada dia.

Use o seguinte array para sua análise:

```python
vendas = np.array([[50, 60, 70, 65, 80],
                   [85, 90, 78, 92, 88],
                   [72, 75, 68, 77, 76]])
```

**Solução:**


````python
import numpy as np

vendas = np.array([[50, 60, 70, 65, 80],
                   [85, 90, 78, 92, 88],
                   [72, 75, 68, 77, 76]])
````

````python
# Vendas totais produto
vendas_totais_produto = vendas.sum(axis=1)
print(f"Vendas totais para cada produto: {vendas_totais_produto}")

# Outra forma
vendas_totais_produto2 = np.sum(vendas,axis=1)
print(vendas_totais_produto2)

# Escrevendo o nome do produto
print(f"Vendas totais do produto A : {vendas_totais_produto[0]}")
print(f"Vendas totais do produto B : {vendas_totais_produto[1]}")
print(f"Vendas totais do produto C : {vendas_totais_produto[2]}")
````

````python
# Vendas totais dia
vendas_totais_dia = vendas.sum(axis=0)
print(f"Vendas totais por dia {vendas_totais_dia}")

# Outra forma
vendas_totais_dia2 = np.sum(vendas, axis=0)
print(vendas_totais_dia2)

# Escrevendo os dias
for i in range(5):
    print(f"Total de vendas para o dia {i +1} : {vendas_totais_dia[i]}")
````
